Normalized

```python
mean_a = np.mean(a)
std_a = np.std(a)

norm_a = (a - mean_a) / std_a
mean_a, std_a, norm_a
```

```python
np.mean(norm_a)
np.std(norm_a)
```

```python
b = np.array(
    [
        [1, 2, 3, 4],
        [5, 6, 7, 8]
    ]
)

mean_b = np.mean(b)
std_b = np.std(b)

norm_b = (b - mean_b) / std_b

mean_b, std_b, norm_b

```

```python
np.mean(norm_b)
```

```python
np.std(norm_b)
```
