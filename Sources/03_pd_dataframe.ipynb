دسترسی به سطر  

iloc: دسترسی با شماره‌های عددی سطر و ستون  

loc: به جای عدد بهش نام سطر و ستون رو میدیم

```python
df.iloc[1:, :2]
df.iloc[1:, -2:]
df.iloc[[True, False, True, False]]
```

```python
df.loc[:3, ["Address", "Qualification"]]
df.loc[:3, "Address":"Qualification"]
df.loc[[0, 2], ["Address", "Qualification"]]
df.loc[[True, False, True, False], ["Qualification"]]
```
