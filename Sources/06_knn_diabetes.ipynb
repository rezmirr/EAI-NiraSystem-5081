```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, recall_score, precision_score
# import matplotlib.pyplot as plt
```

```python
data_set = pd.read_csv(".\DB\diabetes.csv")
data_set.head()
```

اصلاح missing values

```python
zero_not_accepted = [
    "Glucose",
    "BloodPressure",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "Age",
]
```

```python
for feature in zero_not_accepted:
    data_set[feature] = data_set[feature].replace(0, np.nan)
    mean = data_set[feature].mean(skipna=True)
    data_set[feature] = data_set[feature].replace(np.nan, mean)
```

```python
data_set["Insulin"]
```

جداسازی داده‌ها

```python
x = data_set.iloc[:, :-1]
y = data_set.iloc[:, -1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
knn_model = KNeighborsClassifier(n_neighbors=11)
knn_model.fit(x_train, y_train)
y_pred = knn_model.predict(x_test)
```

معیارهای ارزیابی

```python
acc = accuracy_score(y_test, y_pred)

print(f"ACC: {acc * 100:.2f}%")
```
