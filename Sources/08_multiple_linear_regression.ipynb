```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
dataset = pd.read_csv("DB\petrol_consumption.csv")
```

جداسازی داده ها

```python
x = dataset[[
    "Petrol_tax",
    "Average_income",
    "Paved_Highways",
    "Population_Driver_licence(%)",
]]

y = dataset["Petrol_Consumption"]

x.shape, y.shape
```

```python
x_train, x_test, y_trian, y_test = train_test_split(x, y, random_state=40)
```

```python
regressor_model = LinearRegression()
regressor_model.fit(x_train, y_trian)

regressor_model.coef_, regressor_model.intercept_
```

```python
coef_df = pd.DataFrame(
    regressor_model.coef_, x.columns, columns=["Coefficient"]
)

coef_df
```

```python
y_pred = regressor_model.predict(x_test)

df = pd.DataFrame({
    "Actual": y_test,
    "Predict": y_pred
})

df
```

```python
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)

print(f"MAE: {mae:.2f}, MSE: {mse:.2f}, RMSE: {rmse:.2f}")
```
